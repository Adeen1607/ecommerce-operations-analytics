# E-commerce Operations — Key Visuals

Run the marketplace model first. The notebook reviews delivery reliability, freight burden, seller concentration, and review outcomes.


```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

DATA = Path("../data/processed")
orders = pd.read_csv(DATA / "fact_orders.csv", parse_dates=["order_purchase_timestamp"])
monthly = pd.read_csv(DATA / "monthly_marketplace_kpis.csv")
sellers = pd.read_csv(DATA / "seller_scorecard.csv")
```

```python
monthly["purchase_month"] = pd.to_datetime(monthly["purchase_month"])
figure, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(monthly["purchase_month"], monthly["product_revenue"], color="#2563EB")
axes[0].set(title="Monthly Product Revenue", ylabel="Revenue")
axes[1].plot(monthly["purchase_month"], monthly["on_time_delivery_rate"], color="#059669")
axes[1].set(title="On-time Delivery Rate", ylabel="Rate", ylim=(0, 1))
for axis in axes:
    axis.grid(alpha=0.25)
figure.tight_layout()
```

```python
review_delivery = orders.dropna(subset=["review_score", "delivery_variance_days"])
figure, axis = plt.subplots(figsize=(8, 5))
review_delivery.boxplot(column="delivery_variance_days", by="review_score", ax=axis)
axis.set(title="Delivery Variance by Review Score", xlabel="Review score", ylabel="Days vs estimate")
figure.suptitle("")
figure.tight_layout()
```

```python
top = sellers.head(15).sort_values("product_revenue")
figure, axis = plt.subplots(figsize=(10, 7))
axis.barh(top["seller_id"], top["product_revenue"], color="#7C3AED")
axis.set(title="Top Sellers by Product Revenue", xlabel="Revenue")
figure.tight_layout()
```
